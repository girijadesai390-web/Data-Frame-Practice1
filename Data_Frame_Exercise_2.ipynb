```python
def temp_categories(temp):
  if temp >-5 and temp < 31:
    return "Ideal Range"
  elif temp > 30 and temp < 51:
    return "Normal Operating Range"
  elif temp > 50 and temp < 71:
    return "Upper normal range"
  else:
    return "High range"
```

```python
df["Temperature_Categories"] = df.apply(lambda x: temp_categories(x["Temperature"]), axis=1)
df
```

```python
def Holiday_categories(Holiday):
  if Holiday==0:
    return"NO"
  else:
    return"YES"
```

```python
df["Holiday_Categories"] = df.apply(lambda x: Holiday_categories(x["Holiday_Flag"]), axis=1)
df
```

```python
import matplotlib.pyplot as plt
import numpy as np
np.random.seed(42)
study_hours=np.random.uniform(1,10,size=50)
exam_scores=40+(study_hours*5.5)+np.random.normal(0,5,size=50)
plt.scatter(study_hours,exam_scores,color="red")
plt.xlabel="Study Hours"
plt.ylabel="Exam Scores"
plt.title("Scatter Plot")
plt.show()
```
